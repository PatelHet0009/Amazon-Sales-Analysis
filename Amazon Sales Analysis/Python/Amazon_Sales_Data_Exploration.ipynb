```python
customer_orders = dataFile.groupby('Customer Name')['Order ID'].count().sort_values(ascending=False)
print(customer_orders)
```

Payment Method Analysis

```python
payment_orders = dataFile.groupby('Payment Method')['Order ID'].count().sort_values(ascending=False)
print(payment_orders)
```

```python
orders_status = dataFile.groupby('Status')['Order ID'].count().sort_values(ascending=False)
print(orders_status)
```

```python
orders_revenue = dataFile.groupby('Status')['Total Sales'].sum().sort_values(ascending=False)
print(orders_revenue)
```

```python
total_orders = len(dataFile)
cancelled_orders = (dataFile['Status'] == 'Cancelled').sum()
cancellation_rate = (cancelled_orders / total_orders) * 100
print(f"Cancellation Rate: {cancellation_rate:.2f}%")
```

```python
plt.figure(figsize=(8, 5))
bars = plt.bar(orders_status.index, orders_status.values)
plt.title('Order Status Distribution')
plt.xlabel('Order Status')
plt.ylabel('Number of Orders')
for bar in bars:
    value = bar.get_height()
    percentage = (value / total_orders) * 100
    
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f'{int(value)} ({percentage:.1f}%)',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

### Business Insight

- Completed orders account for 35.2% of all orders.
- Pending orders account for 34.0% of all orders.
- Cancelled orders account for 30.8% of all orders.
- The cancellation rate of 30.8% is relatively high within this dataset.
- Pending orders also represent a substantial portion of total orders, indicating that order processing or fulfillment could be investigated further.

```python
# Revenue Contribution by Product within Category

category_product_pivot = dataFile.pivot_table(
    index='Category',
    columns='Product',
    values='Total Sales',
    aggfunc='sum',
    fill_value=0
)

ax = category_product_pivot.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6)
)

plt.title('Revenue Contribution by Product within Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue (₹)')
plt.xticks(rotation=0)

# Add some space above the highest bar
plt.ylim(0, category_product_pivot.sum(axis=1).max() * 1.15)

plt.legend(
    title='Product',
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

# Key Business Insights

### 1. Overall Sales Performance
The dataset contains 250 orders with total sales of ₹243,845 and 714 units sold. The average order value is ₹975.38.

### 2. Category Performance
Electronics is the highest-performing category with ₹129,950 in sales, followed by Home Appliances with ₹105,000. Together, these two categories contribute approximately 96.4% of total sales.

### 3. Product Performance
Refrigerator is the highest-revenue product, generating ₹78,000. Laptop and Smartphone follow with ₹58,400 and ₹48,500 respectively.

### 4. Quantity vs Revenue
Smartwatch recorded the highest quantity sold at 105 units, while Refrigerator generated the highest revenue with 65 units sold. This indicates that higher sales volume does not necessarily result in higher revenue because product prices differ.

### 5. Customer Performance
Olivia Wilson generated the highest customer revenue at ₹36,170 and also recorded the highest average order value of approximately ₹1,247. Emma Clark placed the highest number of orders with 32 orders.

### 6. Payment Method Performance
PayPal recorded the highest number of orders (60), highest sales value (₹69,645), and highest average order value (₹1,160.75) among the available payment methods.

### 7. Order Status
There were 88 completed orders, 85 pending orders, and 77 cancelled orders. The cancellation rate is 30.80%, indicating that cancellations represent a significant portion of the orders and may require further investigation.

### 8. Monthly Sales
February generated ₹122,695 in sales, while March generated ₹117,730. April recorded only ₹3,420; however, the dataset contains only two dates in April, so the April value should not be interpreted as a complete-month decline.
